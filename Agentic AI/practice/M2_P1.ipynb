# Reflection Pattern Practice: Python Function Generator

This practice notebook implements the **Reflection Design Pattern** that I learned in the Agentic AI course. This is a simplified, text-only exercise to reinforce my understanding of how reflection works.

## What You'll Build

In this notebook, I will implement a **two-step reflection workflow** that:

1. **Generates a Python function (V1)** based on a user requirement
2. **Reflects on the code quality** by analyzing correctness, edge cases, readability, documentation, and efficiency
3. **Generates an improved function (V2)** that addresses all the feedback

Unlike the chart generation lab (which used multi-modal LLMs to analyze images), this exercise focuses purely on **text-based reflection** — making it simpler to understand the core pattern.

## Workflow Overview

```
User Requirement
      ↓
[Generate V1 Code] → Draft function (JSON output)
      ↓
[Reflect on V1] → Analyze code quality (JSON with feedback)
      ↓
[Generate V2 Code] → Improved function (JSON output)
```

## Example Use Case

**Requirement:** "Write a function that calculates the factorial of a number"

**What happens:**
- **V1**: Basic factorial function (may lack error handling, docstrings, type hints)
- **Reflection**: LLM identifies missing edge cases (negative numbers, zero), lack of documentation, and suggests improvements
- **V2**: Production-ready function with docstring, type hints, proper error handling, and clear variable names

## Why This Pattern Matters

The reflection pattern helps:
- Improve code quality automatically
- Catch edge cases and bugs early
- Generate production-ready code, not just working code
- Demonstrate how LLMs can act as code reviewers

# 1. Initialize Environment & Variables

````python
# Imports
import json
import utils

# Variables
initial_model = "Azure-GPT-4.1"
reflection_model = "Azure-GPT-4o"
requirements = "Create a Python function that takes a list of numbers and returns the sum of all even numbers in the list."
````

# 2. Building the Pipeline

## 2.1 Generate Initial Code (V1)

````python
def generate_initial_code_v1(model: str, requirements: str) -> str:
    """
    Generates initial code based on the provided requirements using a language model.
    """

    prompt = f"""
    You are an expert Python developer.

    Write a Python function based on the following requirement:
    {requirements}

    Return your response ONLY as valid JSON in this exact format:
    {{
    "content": "<complete Python function code here>"
    }}

    Requirements:
    - Write clean, working Python code
    - Include the function definition only
    - Do not add explanations, comments, or markdown
    - Ensure the JSON is valid and parseable
    """

    response = utils.get_response(model=model, prompt=prompt)
    response_json = json.loads(response)
    return response_json["content"]
````

Now we will try out the first step of our pipeline: generating the initial version of the Python function based on the user requirement.

````python
# Generate initial code example
initial_code = generate_initial_code_v1(initial_model, requirements)
print(initial_code)
````

# 3. Reflect on the Initial Code (V1)

````python
def reflect_on_code_v1(model: str, code: str, requirements: str) -> tuple[str, str]:
    """
    Reflects on the provided code and suggests improvements based on the requirements.
    """

    prompt = f"""
    You are a senior code reviewer specializing in Python best practices.

    Review the following Python function code and provide constructive feedback.

    Original requirement: {requirements}

    Code to review:
    {code}

    Analyze the code for:
    1. **Correctness**: Does it meet the requirement? Any logic errors?
    2. **Edge cases**: Are there missing validations or error handling?
    3. **Readability**: Are variable names clear? Is the logic easy to follow?
    4. **Documentation**: Does it need a docstring or comments?
    5. **Efficiency**: Can it be optimized?

    Return your response ONLY as valid JSON in this exact format:
    {{
    "feedback": "<detailed critique covering the 5 points above>",
    "content": "<improved Python function code with all fixes applied>"
    }}

    Requirements for improved code:
    - Add proper docstring
    - Handle edge cases
    - Use clear variable names
    - Include type hints if appropriate
    - Ensure it's production-ready
    - Do not add markdown formatting
    """

    response = utils.get_response(model=model, prompt=prompt)
    response_json = json.loads(response)
    return (response_json["feedback"], response_json["content"])
````

Now we will try out the second step of our pipeline: reflecting on the initial version of the Python function to identify areas for improvement.

````python
# Generate feedback alongside reflected code
feedback, improved_code = reflect_on_code_v1(
    reflection_model, initial_code, requirements
)
print("Feedback:\n", feedback)
print("Improved Code:\n", improved_code)
````

# 4. Putting It All Together - creating the end-to-end workflow

````python
def run_workflow(
    model_initial: str, model_reflect: str, requirements: str
) -> tuple[str, str, str]:
    """
    Runs the end-to-end workflow of generating initial code and reflecting on it.
    Returns the initial code, feedback, and improved code.
    """

    # Step 1: Generate initial code
    initial_code = generate_initial_code_v1(model_initial, requirements)

    # Step 2: Reflect on the initial code
    feedback, improved_code = reflect_on_code_v1(
        model_reflect, initial_code, requirements
    )

    return (initial_code, feedback, improved_code)
````

Now we will try the entire reflection workflow end-to-end, generating V1 code, reflecting on it, and producing V2 code.

````python
# Run the complete workflow

initial_code, feedback, improved_code = run_workflow(
    initial_model, reflection_model, requirements
)

print("Initial Code:\n", initial_code)
print("Feedback:\n", feedback)
print("Improved Code:\n", improved_code)
````
